# ViviendaSeg — Validación automática Python ↔ Prolog

**Objetivo:** reproducir la validación de las reglas SI–ENTONCES y comparar ambos motores sin instalar SWI-Prolog en Windows.

1. En Google Colab, selecciona **Entorno de ejecución → Ejecutar todas**.
2. Autoriza la ejecución del cuaderno cuando Colab lo solicite.
3. Las celdas instalarán SWI-Prolog **solo en el entorno temporal de Colab**, descargarán el repositorio público y ejecutarán las pruebas.

**Criterios de éxito:** 16 pruebas unitarias Prolog correctas y 514 escenarios de comparación Python–Prolog sin diferencias.

**Aviso:** es un prototipo educativo sin validación actuarial ni pericial; la web de Cloudflare no se modifica.

**Diagnóstico:** Si la comparación falla, la tercera celda muestra directamente el escenario y el error, sin ocultarlos tras un `CalledProcessError`.


```python
import pathlib, subprocess, shutil

def ejecutar(cmd, cwd=None):
    print('> ' + ' '.join(cmd), flush=True)
    subprocess.run(cmd, cwd=cwd, check=True)

print('Preparando SWI-Prolog en el entorno temporal de Colab...')
if shutil.which('swipl') is None:
    ejecutar(['apt-get', 'update', '-qq'])
    ejecutar(['apt-get', 'install', '-y', '-qq', 'swi-prolog-nox'])
ejecutar(['swipl', '--version'])

carpeta = pathlib.Path('/content/ViviendaSeg-Sistema-Experto')
if carpeta.exists():
    ejecutar(['git', 'pull', '--ff-only'], cwd=carpeta)
else:
    ejecutar(['git', 'clone', '--depth', '1', 'https://github.com/raineragoge/ViviendaSeg-Sistema-Experto.git', str(carpeta)])
print('\nRepositorio listo en:', carpeta)
```

```python
import sys, subprocess
print('1/2 — Comprobando las 16 pruebas unitarias de Prolog', flush=True)
pruebas = subprocess.run(
    ['swipl', '-q', '-s', 'tests/test_rules.pl', '-g', '(run_tests -> halt(0); halt(1))', '-t', 'halt'],
    cwd=carpeta, capture_output=True, text=True)
print('SALIDA PROLOG:\n' + (pruebas.stdout or '(sin salida estándar)'), flush=True)
print('DETALLE PROLOG:\n' + (pruebas.stderr or '(sin avisos)'), flush=True)
print('Código de salida de las pruebas Prolog:', pruebas.returncode, flush=True)

if pruebas.returncode != 0:
    print('❌ Las pruebas Prolog NO se han superado. Copia todo el bloque SALIDA / DETALLE.', flush=True)
else:
    print('\n2/2 — Comparando 514 escenarios Python–Prolog', flush=True)
    comparacion = subprocess.run(
        [sys.executable, '-u', 'scripts/test_prolog_parity.py'],
        cwd=carpeta, capture_output=True, text=True)
    print('SALIDA DEL COMPARADOR:\n' + (comparacion.stdout or '(sin salida estándar)'), flush=True)
    print('ERRORES DEL COMPARADOR:\n' + (comparacion.stderr or '(sin errores impresos)'), flush=True)
    print('Código de salida del comparador:', comparacion.returncode, flush=True)
    if comparacion.returncode == 0:
        print('\n✅ VALIDACIÓN PYTHON–PROLOG SUPERADA: 16 pruebas y 514 escenarios coincidentes.')
    else:
        print('\n❌ COMPARACIÓN PENDIENTE: copia las líneas bajo SALIDA DEL COMPARADOR y ERRORES DEL COMPARADOR para identificar la discrepancia.')
```

## Interpretación de los resultados

- Si aparece **VALIDACIÓN PYTHON–PROLOG SUPERADA**, los resultados comparados coinciden.
- Si aparece **DIFERENCIA en escenario N**, copia el número de escenario y los datos mostrados para corregir la discrepancia.
- Si una prueba Prolog falla, copia el mensaje de error completo. No des por terminada la validación técnica hasta corregirlo.

Las pruebas comprueban reglas de un sistema experto académico, no la seguridad real de una vivienda.
